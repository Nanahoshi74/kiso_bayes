```python
import math
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib
```

```python
np.random.seed(0)
```

```python
alpha = 11
lam = 13

epsilon = 0.05
L = 15

theta0 = 0.1
p0 = 0.0
```

# リープフロッグでの遷移

```python
# 式5.51
def h(theta):
	return lam * theta - (alpha-1) * math.log(theta)

# 式5.52
def dh(theta):
	return lam - (alpha - 1) / theta

# ハミルトニアン
def H(theta, p):
	return h(theta) + p**2 / 2

def leapfrog_step(theta, p, epsilon):
	# 運動量を半ステップ進める。式5.31
	p_half = p - epsilon / 2 * dh(theta)

	# 位置を1ステップ進める。式5.32
	theta_new = theta + epsilon * p_half

	# 運動量をさらに半ステップ進める。式5.33
	p_new = p_half - epsilon / 2 * dh(theta_new)

	return theta_new, p_new

theta = theta0
p = p0

print(
	f"{'tau':>3} "
	f"{'p':>8} "
	f"{'theta':>8} "
	f"{'h(theta)':>10} "
	f"{'H(theta,p)':>12}"
)
for tau in range(1, L + 1):
	print(
		f"{tau:3d} "
		f"{p:8.2f} "
		f"{theta:8.2f} "
		f"{h(theta):10.2f} "
		f"{H(theta, p):12.2f}"
	)
	# リープフロッグ法で更新
	if tau < L:
		theta, p = leapfrog_step(theta, p, epsilon)
```

# HMC

```python
def hmc_step(theta_current):
	# 運動量を標準正規分布から発生
	p_current = np.random.normal(0.0, 1.0)

	theta = theta_current
	p = p_current
	H_current = H(theta, p)

	# リープフロッグで更新
	for _ in range(L):
		theta, p = leapfrog_step(theta, p, epsilon)

	theta_prop = theta
	p_prop = p

	# 提案状態でのハミルトニアン
	H_prop = H(theta_prop, p_prop)

	# メトロポリス受容確率
	accept_prob = min(1, np.exp(H_current - H_prop))

	u = np.random.uniform(0.0, 1.0)

	if u < accept_prob:
		return theta_prop, True
	else:
		return theta_current, False
```

# HMC の実行

```python
# HMC の実行回数
T = 100

theta = 2.5

# 各時点で新しく発生させた p^(t)
start_momenta = []

# 各HMC遷移のリープフロッグ軌跡
trajectories = []

# リープフロッグ終了地点
end_points = []
theta_samples = [theta]
accepted = 0

for t in range(T):
	theta, accept = hmc_step(theta)
	theta_samples.append(theta)

	if accept:
		accepted += 1

print("受容数 =", accepted)
print("受容率 =", accepted / T)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
	range(T + 1),
	theta_samples,
	marker="o",
	markersize=4
)

plt.xlabel("t")
plt.ylabel(r"$\theta$")
plt.title("トレースプロット")

plt.grid()

plt.show()
```

```python
burn_in = 100
samples = np.array(theta_samples)
samples_after_burnin = samples[burn_in:]
mean_theta = np.mean(samples_after_burnin)

print("バーンイン =", burn_in)
print("バーンイン後のサンプル数 =", len(samples_after_burnin))
print("theta の平均 =", mean_theta)
```

```python
np.random.seed(0)

epsilon_plot = 0.01
L_plot = 10000
T_plot = 10000

theta = 2.5
theta_plot = [theta]
start_momenta = []
trajectories = []
end_points = []

for t in range(T_plot - 1):
	p_current = np.random.normal(0.0, 1.0)
	start_momenta.append(p_current)

	theta_work = theta
	p_work = p_current
	H_current = H(theta_work, p_work)

	theta_path = [theta_work]
	p_path = [p_work]

	for _ in range(L_plot):
		theta_work, p_work = leapfrog_step(theta_work, p_work, epsilon_plot)
		theta_path.append(theta_work)
		p_path.append(p_work)

	theta_prop = theta_work
	p_prop = p_work
	H_prop = H(theta_prop, p_prop)

	trajectories.append((np.array(theta_path), np.array(p_path)))
	end_points.append((theta_prop, p_prop))

	accept_prob = min(1.0, np.exp(H_current - H_prop))

	if np.random.uniform(0.0, 1.0) < accept_prob:
		theta = theta_prop

	theta_plot.append(theta)

theta_plot = np.array(theta_plot)
```

```python
start_momenta.append(
	np.random.normal(0.0, 1.0)
)

start_momenta = np.array(start_momenta)

p_grid = np.linspace(-5, 5, 400)
theta_grid = np.linspace(0.05, 3.2, 400)

P, THETA = np.meshgrid(
	p_grid,
	theta_grid
)

H_grid = (
	lam * THETA
	- (alpha - 1) * np.log(THETA)
	+ P**2 / 2
)

levels = np.arange(14, 32, 2)
```

```python
plt.figure(figsize=(7, 7))


# ハミルトニアンの等高線
cs = plt.contour(
	P,
	THETA,
	H_grid,
	levels=levels
)

plt.clabel(
	cs,
	inline=True,
	fontsize=8
)

for t in range(len(trajectories)):

	theta_path, p_path = trajectories[t]

	plt.plot(
		p_path,
		theta_path,
		linewidth=0.6
	)


# ---------------------------------
# 各 t で運動量を発生させた点
# ---------------------------------
plt.scatter(
	start_momenta,
	theta_plot,
	s=12,
	facecolors="none"
)


# 各時点を線で結ぶ
plt.plot(
	start_momenta,
	theta_plot,
	linewidth=0.5
)


plt.xlim(-5, 5)
plt.ylim(0, 3.2)

plt.xlabel("p")
plt.ylabel(r"$\theta$")

plt.title("Phase space transition: t = 1 ... 100")

plt.grid()

plt.show()
```
