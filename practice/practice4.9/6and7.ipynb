```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# mu_q = 0
# var_q = 0.5

mu_q = 3
var_q = 0.5

sd_q = np.sqrt(var_q)

alpha = 11
lam = 13
```

```python
def f(theta):
	if theta <= 0:
		return 0
	return np.exp(-13*theta) * theta**10

def q(theta):
	return norm.pdf(theta, loc=mu_q, scale=sd_q)
```

```python
n_list = [10, 100, 1000, 10000, 100000]
accepted = 0


for n in n_list:
	theta = np.zeros(n + 1)
	theta[0] = 1.0
	burn_in = n // 100
	for t in range(n):
		a = np.random.normal(mu_q, sd_q)
		r = (q(theta[t]) * f(a)) / (q(a) * f(theta[t]))

		if np.random.rand() < min(1, r):
			theta[t + 1] = a
			accepted += 1
		else:
			theta[t + 1] = theta[t]
	samples = theta[burn_in + 1:]
	print(f"------------------ n = {n}------------------")
	print(f"受容率: {accepted / n}")
	print(f"MCMC平均: {np.mean(samples)}")
	print(f"MCMC分散: {np.var(samples)}")

print(f"理論値平均: {alpha / lam}")
print(f"理論値分散: {alpha / lam**2}")
		
```
