```python
import numpy as np
rng = np.random.default_rng()
```

```python
n_list = [10, 100, 1000, 10000, 100000]

for n in n_list:
	x = []
	for i in range(n):
		x1 = rng.uniform(0, 1)
		x2 = rng.uniform(0, 1)

		if x1**2 + x2**2 < 1:
			x.append(4)
		else:
			x.append(0)

	print(f"n = {n}, mean = {np.sum(x) / n}")
	
```
